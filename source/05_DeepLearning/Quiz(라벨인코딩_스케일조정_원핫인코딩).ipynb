- 종속변수(y) : 평당분양가격

# 5. 지역명을 원핫인코딩

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

# 1. 데이터 로드
url = 'https://raw.githubusercontent.com/4aix/data/refs/heads/master/ch13_apt_fillna_median.csv'
df = pd.read_csv(url)

# 2. 지역명 라벨인코딩 (지역명2 생성)
le = LabelEncoder()
df['지역명2'] = le.fit_transform(df['지역명'].astype(str))

# 스케일링 대상 설정
X = df[['지역명2', '연도', '월']]
y = df[['평당분양가격']]

# 3. MinMaxScaling (정규화: 0~1 변환)
mm_scaler = MinMaxScaler()
X_m = mm_scaler.fit_transform(X)
y_m = mm_scaler.fit_transform(y)

df_m = pd.DataFrame(X_m, columns=['지역명2m', '연도m', '월m'])
df_m['평당분양가격m'] = y_m

# 4. StandardScaling (표준화: 평균0, 표준편차1 변환)
std_scaler = StandardScaler()
X_s = std_scaler.fit_transform(X)
y_s = std_scaler.fit_transform(y)

df_s = pd.DataFrame(X_s, columns=['지역명2s', '연도s', '월s'])
df_s['평당분양가격s'] = y_s

# 5. 지역명 원핫인코딩 (One-Hot Encoding)
df_ohe = pd.get_dummies(df['지역명'], prefix='지역명', dtype=int)

# 6. 원본 및 전처리 데이터프레임 최종 병합
df_base = df[['지역명', '연도', '월', '지역명2', '평당분양가격']]
df_final = pd.concat([df_base, df_m, df_s, df_ohe], axis=1)

# 데이터프레임 출력 확인
df_final
```
