```python
import tkinter as tk
import random

class Minesweeper:
    def __init__(self, root, size=8, mines=10):
        self.root = root
        self.root.title("파이썬 지뢰찾기")
        self.size = size
        self.mines = mines
        
        self.buttons = {}
        self.mine_positions = set()
        self.create_board()
        self.place_mines()

    def create_board(self):
        for r in range(self.size):
            for c in range(self.size):
                btn = tk.Button(self.root, text="", width=4, height=2, font=('Arial', 12, 'bold'),
                                command=lambda r=r, c=c: self.click(r, c))
                btn.grid(row=r, column=c)
                self.buttons[(r, c)] = btn

    def place_mines(self):
        positions = [(r, c) for r in range(self.size) for c in range(self.size)]
        self.mine_positions = set(random.sample(positions, self.mines))

    def click(self, r, c):
        if (r, c) in self.mine_positions:
            self.buttons[(r, c)].config(text="💣", bg="red")
            self.game_over(False)
        else:
            self.reveal(r, c)

    def reveal(self, r, c):
        if self.buttons[(r, c)]['state'] == 'disabled':
            return

        # 주변 지뢰 개수 계산
        count = sum((nr, nc) in self.mine_positions 
                    for nr in range(r-1, r+2) for nc in range(c-1, c+2)
                    if 0 <= nr < self.size and 0 <= nc < self.size)

        self.buttons[(r, c)].config(text=str(count) if count > 0 else "", state='disabled', relief=tk.SUNKEN, bg="lightgray")

        # 주변에 지뢰가 없으면 자동 연쇄 오픈
        if count == 0:
            for nr in range(r-1, r+2):
                for nc in range(c-1, c+2):
                    if 0 <= nr < self.size and 0 <= nc < self.size and (nr, nc) != (r, c):
                        if self.buttons[(nr, nc)]['state'] != 'disabled':
                            self.reveal(nr, nc)

    def game_over(self, won):
        for (r, c), btn in self.buttons.items():
            if (r, c) in self.mine_positions:
                btn.config(text="💣", bg="red")
            btn.config(state='disabled')
        
        msg = "축하합니다! 승리하셨습니다!" if won else "지뢰를踏았습니다! GAME OVER"
        tk.Label(self.root, text=msg, font=('Arial', 14, 'bold')).grid(row=self.size, column=0, columnspan=self.size)

if __name__ == "__main__":
    root = tk.Tk()
    game = Minesweeper(root)
    root.mainloop()
```
