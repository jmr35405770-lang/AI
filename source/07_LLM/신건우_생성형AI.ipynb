```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatOllama(model='llama3.2:1b')
```

```python
food_template = PromptTemplate(template='What is the famous food {country} ? Return the name of the food only.',
                               input_variables=['country'])

output_parser = StrOutputParser()
    
first_chain = food_template | llm | output_parser
country=input("나라 이름을 입력하세요 : ")
first_chain.invoke({'country':country})
```

```python
recipe_template = PromptTemplate(template='What is the famous recipe for {food} ?',
                               input_variables=['food'])

output_parser = StrOutputParser()
    
second_chain = recipe_template | llm | output_parser
food=input("음식 이름을 입력하세요 : ")
second_chain.invoke({'food':food})
```

```python
final_chain = first_chain | second_chain 
```

```python
result = final_chain.invoke({'country': 'korea'})
```
